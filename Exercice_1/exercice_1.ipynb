# Introduction
Entrons directement dans le vif du sujet. En tant que développeur IA, tu auras constamment besoin de données pour entraîner tes modèles. <br>Le SQL (Structured Query Language) est l'outil indispensable pour extraire et manipuler ces données.

**Définition courte** : Le SQL est un langage qui permet de poser des questions à une base de données pour y lire, ajouter, modifier ou supprimer des informations.

**Concrètement** : Imagine une base de données comme un grand classeur Excel, et une "Table" comme l'une de ses feuilles de calcul, organisée en lignes et en colonnes.

Voici un exemple de table nommée *Utilisateurs* :

| ID       | Nom      |     Age  |Profession|
|:-------- |:-------- |:--------:| :--------|
| 1        | Alice    | 25       | Data Scientist       |
| 2        | Bob      | 30       | Développeur       |
| 3        | Charlie  | 22       | Etudiant       |

La commande la plus basique en SQL sert à lire des données. Elle se construit avec deux mots-clés principaux :
* SELECT *(Sélectionner)* : indique les colonnes que l'on veut récupérer.

* FROM *(Depuis)* : indique la table où se trouvent ces colonnes.

**Exemple :**
Si je veux récupérer uniquement les noms de tous les utilisateurs, la requête est :<br><br>
<span style="padding:10px; background-color:black;">SELECT Nom FROM Utilisateurs;</span>

Pour faire communiquer Python et une base de données, on utilise des bibliothèques. En IA, la plus incontournable pour manipuler des données sous forme de tableau s'appelle *pandas*.<br>

**Définition courte** : *pandas* est une **bibliothèque Python** qui permet d'analyser et de manipuler des données. Sa fonction *read_sql_query* permet d'exécuter une requête SQL et de transformer le résultat en un tableau interactif (appelé DataFrame).

**Concrètement** : Voici comment on utiliserait le SQL dans Python pour récupérer uniquement les noms :

<span style="color:red;">**ATTENTION** : ne pas oublier d'installer les bibliothèques avant</span> <br>
Où trouver la commande ? [PyPI](https://pypi.org/) Écrivez le nom de la bibliothèques et copier la commande *(exemple : pip install pandas)*

```python
# En premier lieu toujours importer les bibliothèques nécessaires au fonctionnement de ton projet
# import bibliothèque
# Pour une utilisation plus "courte" via un alias :
# import bibliothèque as bib
import sqlite3
# Si pandas n'est pas installé : pip install pandas
import pandas as pd
```

```python
# 1. On simule une connexion à la base de données
connect = sqlite3.connect("ma_base_de_donnees.db")

# 2. On écrit notre requête SQL sous forme de texte
query = "SELECT Nom FROM Utilisateurs"

# 3. Pandas lit la requête et stocke le résultat dans la variable 'donnees'
data = pd.read_sql_query(query, connect)

# On affiche les données
print(data)
```
