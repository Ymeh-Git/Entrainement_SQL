# Introduction

Passons à l'étape suivante. En intelligence artificielle, nous avons des bases de données massives, mais nous voulons entraîner notre modèle uniquement sur une partie spécifique de ces données. Il faut donc **filtrer** les informations.

**Définition courte** : 
En SQL, on utilise le mot-clé WHERE (Où / Sous condition que) à la fin de la requête pour filtrer les résultats selon un critère précis.

**Concrètement** :
Si nous reprenons notre table Utilisateurs (qui contient les colonnes *Nom, Age, Profession*), imaginons que nous voulons récupérer uniquement les noms des personnes qui ont strictement plus de 24 ans.

Pour cela, on utilise des **opérateurs mathématiques** classiques :

* \> (strictement supérieur)

* < (strictement inférieur

* = (égal)

* \>= ou <= (supérieur/inférieur ou égal)

**Exemple :**<br>
SELECT Nom FROM Utilisateurs WHERE Age > 24;

## Exercice en code (Python)

En Python, l'intégration de cette condition ne change pas la structure de notre code. Nous écrivons toujours la requête sous forme de texte avant de demander à pandas de l'exécuter.

```python
# Quelles bibliothèques importer ? Après avoir importer les bibliothèques lancez ce bloc.
import sqlite3
import pandas as pd
```

> ***Astuce*** : <br>en SQL, si l'on veut récupérer toutes les colonnes d'une table sans avoir à les lister une par une, on utilise le symbole astérisque **'*'**.<br>
Par exemple : SELECT * FROM Utilisateurs; <span style="color:green;"># récupère tout le tableau.</span>

Si tu dois écrire le code Python pour définir une variable nommée **query** contenant la requête SQL qui récupère **toutes les colonnes** de la table *Performances_IA*, mais uniquement pour les modèles dont le **Temps_Entrainement** est **strictement inférieur à 50**, comment rédiges-tu cette ligne de code ?

```python
connect = sqlite3.connect("my_database.db")

query = "SELECT * FROM Performances_IA WHERE Temps_Entrainement < 50"

df = pd.read_sql_query(query, connect)
```
