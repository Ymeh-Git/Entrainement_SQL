# Introduction

Passons à une notion indispensable lorsque l'on prépare des données pour entraîner ou évaluer une IA : l'**agrégation**. Au lieu de regarder chaque ligne individuellement, il est souvent nécessaire d'obtenir des statistiques globales.

**Définition courte** : L'**agrégation** SQL permet de rassembler des lignes qui partagent une même valeur (avec le mot-clé **GROUP BY**) pour leur appliquer un calcul statistique comme :
* la **moyenne (AVG)**,
* le **total (SUM)**,
* compter le **nombre d'éléments (COUNT)**.

**Formule mathématique** :
La fonction SQL la plus utilisée en IA est **AVG()** pour calculer une moyenne arithmétique. Mathématiquement, la moyenne $\mu$ d'un ensemble de $n$ valeurs s'écrit :

$$\mu = \frac{1}{n} \sum_{i=1}^{n} x_i$$

* $x_i$ représente chaque valeur individuelle (par exemple, la précision de chaque modèle).
* $\sum$ indique qu'il faut additionner toutes ces valeurs.
* $n$ est le nombre total d'éléments.

Pour connaître le score moyen par catégorie, la requête SQL regroupe les catégories et calcule la moyenne des scores pour chacune : <br>

*SELECT Categorie, AVG(Score) FROM Evaluations GROUP BY Categorie;*
