il nous manque une dernière brique SQL fondamentale pour ton métier de développeur IA : Les **Jointures** ( *JOIN* )

En intelligence artificielle, tes données d'entraînement ne sont presque jamais stockées dans un seul tableau.<br>
Par exemple, tu peux avoir les **caractéristiques de tes utilisateurs** dans une table, et leur **historique d'achat** dans une autre. Il faut pouvoir les **fusionner**.

**Définition courte** : La commande **JOIN** permet de combiner les colonnes de deux tables différentes en se basant sur une colonne qu'elles ont en commun *(souvent un identifiant, ou "ID")*.

**Formule mathématique (Théorie des ensembles)** :
La jointure classique (**INNER JOIN**) correspond à l'**intersection de deux ensembles $A$ et $B$**. Mathématiquement, cela s'écrit :

$$A \cap B = \{x \mid x \in A \text{ et } x \in B\}$$

Cela signifie que la requête ne **conservera** que les **lignes où la valeur de la colonne commune existe** à la fois dans la table $A$ et dans la table $B$.

**Exemple concret** :<br>
Imaginons une table Familles (avec un ID_Famille et un Nom_Famille) et une table Enfants (avec un Prenom et un Famille_ID).<br>

Pour lier ces deux tables et savoir dans quelle famille est chaque enfant, la syntaxe est :<br>

SELECT Familles.Nom_Famille, Enfants.Prenom FROM Familles JOIN Enfants ON Familles.ID_Famille = Enfants.Famille_ID;<br>

>Note : on utilise un point "." pour préciser de quelle table vient la colonne.
