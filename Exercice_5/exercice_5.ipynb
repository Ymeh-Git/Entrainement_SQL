Une fois que tu as extrait tes données avec SQL, tu vas te rendre compte d'une réalité du métier : les données du monde réel sont toujours incomplètes. <br>C'est ici qu'intervient le nettoyage des données **(Data Cleaning)**, et plus particulièrement la gestion des valeurs manquantes.

**Définition courte** : Le nettoyage des données consiste à repérer et corriger les "trous" (valeurs absentes, souvent notées *NaN* en Python) dans notre tableau, car un modèle d'IA est incapable de faire des mathématiques avec du vide.

**Concrètement** :<br>
Imagine un algorithme qui prédit le prix d'une maison. Si pour l'une des maisons, le propriétaire a oublié de renseigner la surface, l'IA va planter au moment de l'entraînement.<br>
Deux solutions principales s'offrent à nous :

* **Supprimer** complètement la ligne de cette maison (on **perd de l'information**).

* L'**imputation** : **Remplacer** le vide par une valeur logique, le plus souvent la **moyenne des autres maisons**.

**Formule mathématique** (Imputation par la moyenne) :<br>
Pour remplacer la valeur manquante, on calcule la moyenne $\mu$ uniquement sur les valeurs existantes.

$$\text{Valeur de remplacement} = \mu = \frac{1}{n} \sum_{i=1}^{n} x_i$$

$n$ = le nombre d'éléments qui possèdent bien une valeur.<br>
$\sum_{i=1}^{n} x_i$ = la somme de toutes ces valeurs existantes.

En Intelligence Artificielle, tu vas manipuler des milliers, voire des millions de lignes. Tu ne feras pas ce calcul à la main.

En Python, avec notre fameuse bibliothèque **pandas**, les "**trous**" dans les données sont appelés des **NaN (Not a Number)**.<br>
Pour faire exactement ce que tu viens de faire à la main (l'imputation par la moyenne), pandas nous offre **deux outils très simples** :<br>

* **.mean()** : calcule automatiquement la **moyenne d'une colonne**.

* **.fillna(valeur)** : **remplit** tous les "**trous**" (NaN) d'une colonne avec la **valeur** qu'on lui donne **entre parenthèses**.

**Exemple de syntaxe** :<br>
Si j'ai un tableau nommé donnees et que je veux remplir les trous de la colonne "**Age**" par le chiffre **25**, j'écris :
* donnees["Age"] = donnees["Age"].fillna(25)
