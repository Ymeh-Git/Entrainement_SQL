# Introduction

Maintenant que nos données sont propres et sans trous, il nous reste une dernière étape de préparation très importante avant de donner notre tableau à l'IA : la **mise à l'échelle** (ou Scaling).

**Définition courte** : La normalisation consiste à transformer toutes les données d'une colonne pour qu'elles se situent sur une même échelle, généralement entre 0 et 1.

**Concrètement** :<br>
Reprenons notre IA qui prédit le prix des maisons. Elle se base sur **deux colonnes** : la *surface* (ex: 150 m²) et le nombre de *chambres* (ex: 3).
Si on donne ces chiffres tels quels à l'algorithme, il risque d'accorder beaucoup plus d'importance à la surface simplement parce que le nombre "150" est mathématiquement beaucoup plus grand que "3". L'**IA est biaisée par l'échelle des grandeurs**. Pour éviter cela, on **compresse** toutes les **valeurs entre 0 et 1**.

**Formule mathématique (Min-Max Scaling)** :<br>
Pour convertir une valeur à cette nouvelle échelle, on utilise cette formule :

$$X_{norm} = \frac{X - X_{min}}{X_{max} - X_{min}}$$

$X$ représente la valeur actuelle que l'on veut transformer.<br>
$X_{min}$ est la valeur la plus petite de toute la colonne.<br>
$X_{max}$ est la valeur la plus grande de toute la colonne.

**Définition courte** : scikit-learn est la bibliothèque standard en Python pour le Machine Learning(ML). Elle contient des algorithmes d'IA, mais aussi des outils pour préparer les données, dont le **MinMaxScaler** qui applique exactement la formule que tu viens d'utiliser.

