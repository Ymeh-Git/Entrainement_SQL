# Introduction

Maintenant que nos données sont extraites (SQL), nettoyées (NaN) et mises à l'échelle (Scaling), il reste une toute dernière étape cruciale avant de lancer l'apprentissage de notre modèle d'IA : **La séparation des données** **(Train / Test Split)**.

**Définition courte** : Cela consiste à diviser notre tableau de données en deux morceaux distincts : un gros morceau pour "entraîner" l'IA (Train), et un petit morceau pour la "tester" (Test).

**Concrètement** :<br>
Imagine un étudiant qui prépare le baccalauréat. Pendant l'année, tu lui donnes des exercices pour s'entraîner (le jeu de Train). Mais le jour de l'examen final (le jeu de Test), tu dois lui donner des exercices qu'il n'a jamais vus. S'il avait déjà vu les sujets du bac pendant l'année, il aurait simplement appris les réponses par cœur au lieu de comprendre la logique. Pour l'IA, c'est exactement pareil !

**Formule mathématique** :<br>
En général, on utilise une répartition de 80% pour l'entraînement et 20% pour le test.<br>
Si $N$ est le nombre total de lignes dans ton tableau, la taille des nouveaux tableaux se calcule ainsi :

$$N_{train} = N \times 0.80$$

$$N_{test} = N \times 0.20$$

Exemple :
```Python

# On sépare le tableau en deux.
# test_size=0.20 indique qu'on garde 20% pour le test.
```
